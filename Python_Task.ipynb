Write a Python program to check whether a given number is prime or not

```python
# Store a number
number = 13

# Assume the number is prime
is_prime = number > 1

# Check divisibility
for i in range(2, int(number ** 0.5) + 1):
    if number % i == 0:
        is_prime = False
        break

# Display result
if is_prime:
    print("Prime Number")
else:
    print("Not a Prime Number")
```

Output: Prime Number

Write a recursive Python function to calculate the factorial of a number

```python
# Define recursive factorial function
def factorial(n):
    # Base condition
    if n == 0 or n == 1:
        return 1

    # Recursive call
    return n * factorial(n - 1)

# Call function
print("Factorial:", factorial(5))
```

Output: Factorial: 120

Write a Python function to generate the first seven Fibonacci numbers

```python
# Define Fibonacci function
def fibonacci(count):
    a, b = 0, 1

    # Generate Fibonacci numbers
    for i in range(count):
        print(a, end=" ")
        a, b = b, a + b

# Call the function
fibonacci(7)
```

Output: 0 1 1 2 3 5 8

Write a Python program to create a Student class and display student details using an object

Write a Python program to demonstrate inheritance using parent and child classes

```python
# Create parent class
class Animal:
    def sound(self):
        print("Animals make sounds")

# Create child class
class Dog(Animal):
    def bark(self):
        print("Dog barks")

# Create object
dog = Dog()

# Access parent and child methods
dog.sound()
dog.bark()
```

Output: Animals make sounds
Dog barks

Write a Python program to demonstrate encapsulation using a private variable

```python
# Create a BankAccount class
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    # Access private balance
    def get_balance(self):
        return self.__balance

# Create object
account = BankAccount(5000)

# Display balance
print("Balance:", account.get_balance())
```

Output: Balance: 5000
