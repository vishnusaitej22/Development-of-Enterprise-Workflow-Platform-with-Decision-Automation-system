### Qn29: Calculate factorial using recursion

**Prompt:** Write a recursive function to calculate the factorial of a number.

```python
# Define a recursive factorial function
def factorial(number):
    # Base case stops the recursion
    if number <= 1:
        return 1
    # Recursive case reduces the number
    return number * factorial(number - 1)

# Display the factorial
print("Factorial:", factorial(5))
```

**Logic:** The function calls itself with a smaller value until it reaches the base case of 1.

### Qn30: Generate Fibonacci numbers using a function

**Prompt:** Write a function that returns the first seven numbers in the Fibonacci sequence.

```python
# Define a function to generate Fibonacci numbers
def fibonacci(count):
    sequence = []
    a, b = 0, 1

    for _ in range(count):
        sequence.append(a)
        a, b = b, a + b

    return sequence

# Display the sequence
print("Fibonacci sequence:", fibonacci(7))
```

**Logic:** Each new Fibonacci value is the sum of the previous two values. The function stores the requested number of terms.

### Qn32: Demonstrate inheritance

**Prompt:** Write a Python program in which a Dog class inherits a method from an Animal class.

```python
# Define the parent class
class Animal:
    def eat(self):
        print("Animal is eating")

# Define a child class
class Dog(Animal):
    def bark(self):
        print("Dog is barking")

# Create an object of the child class
dog = Dog()
dog.eat()
dog.bark()
```

### Qn33: Demonstrate polymorphism

```python
# Define two classes with the same method
class Cat:
    def sound(self):
        print("Meow")

class Cow:
    def sound(self):
        print("Moo")

# Call the same method on different objects
for animal in (Cat(), Cow()):
    animal.sound()
```

**Logic:** Both objects provide a sound() method, but each class implements it differently.

### Qn34: Demonstrate encapsulation

**Prompt:** Write a Python program to demonstrate encapsulation using a private balance attribute and a deposit method.

```python
# Define a BankAccount class
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount

    def show_balance(self):
        print("Balance:", self.__balance)

# Create an account and deposit money
account = BankAccount(1000)
account.deposit(500)
account.show_balance()
```

**Logic:** The double underscore makes the balance attribute name-mangled. Public methods provide controlled access to it.
