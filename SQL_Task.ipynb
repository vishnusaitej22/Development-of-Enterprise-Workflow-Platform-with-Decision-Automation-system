### Qn20: Use WHERE and HAVING together

### Qn21: INNER JOIN

**Prompt:** Write an SQL query to display employee names with their matching department names using INNER JOIN.

```python
# Import SQLite library
import sqlite3

# Create a fresh in-memory database for this question
connection = sqlite3.connect(":memory:")

connection.execute("""
CREATE TABLE employees (
    employee_id INTEGER,
    name TEXT,
    department_id INTEGER
)
""")
connection.execute("""
CREATE TABLE departments (
    department_id INTEGER,
    department_name TEXT
)
""")
connection.executemany("INSERT INTO employees VALUES (?, ?, ?)", [
    (101, "Vishnu", 1), (102, "Sai", 2), (103, "Tej", 1), (104, "Raman", None)
])
connection.executemany("INSERT INTO departments VALUES (?, ?)", [
    (1, "Information Technology"), (2, "Human Resources"), (3, "Finance")
])

# Execute the SQL query and display the result
result = connection.execute("""
SELECT employees.name, departments.department_name FROM employees INNER JOIN departments ON employees.department_id = departments.department_id ORDER BY employees.employee_id;
""")
for row in result:
    print(row)

# Close the database connection
connection.close()
```

### Qn25: Join with GROUP BY

```python
# Import SQLite library
import sqlite3

# Create a fresh in-memory database for this question
connection = sqlite3.connect(":memory:")

connection.execute("""
CREATE TABLE employees (
    employee_id INTEGER,
    name TEXT,
    department_id INTEGER
)
""")
connection.execute("""
CREATE TABLE departments (
    department_id INTEGER,
    department_name TEXT
)
""")
connection.executemany("INSERT INTO employees VALUES (?, ?, ?)", [
    (101, "Vishnu", 1), (102, "Sai", 2), (103, "Tej", 1), (104, "Raman", None)
])
connection.executemany("INSERT INTO departments VALUES (?, ?)", [
    (1, "Information Technology"), (2, "Human Resources"), (3, "Finance")
])

# Execute the SQL query and display the result
result = connection.execute("""
SELECT departments.department_name, COUNT(employees.employee_id) AS employee_count FROM departments LEFT JOIN employees ON departments.department_id = employees.department_id GROUP BY departments.department_id, departments.department_name ORDER BY departments.department_id;
""")
for row in result:
    print(row)

# Close the database connection
connection.close()
```
